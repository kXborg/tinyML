```python
import os, tarfile, urllib.request, shutil, random
```

```python
url = "https://storage.googleapis.com/download.tensorflow.org/data/speech_commands_v0.02.tar.gz"
dataset_dir = "data"
os.makedirs(dataset_dir, exist_ok=True)
```

```python
tgz_path = "speech_commands_v0.02.tar.gz"
```

```python
# Extract only yes/no/background folders
wanted = ["yes/", "no/", "_background_noise_/"]
print("Extracting minimal dataset...")
with tarfile.open(tgz_path, "r:gz") as tar:
    for member in tar.getmembers():
        if any(member.name.startswith(w) for w in wanted):
            tar.extract(member, dataset_dir)

# Optional: keep only 100 samples per class
for cmd in ["yes", "no"]:
    cmd_dir = os.path.join(dataset_dir, cmd)
    files = os.listdir(cmd_dir)
    random.shuffle(files)
    for f in files[100:]:
        os.remove(os.path.join(cmd_dir, f))

print("Ready in ./data (yes/no/background)")
```
